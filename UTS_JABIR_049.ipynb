#3. INISIALISASI LIBRARY DAN BROKER


```python
import paho.mqtt.client as mqtt
from time import sleep
from random import randrange

# Inisialisasi broker
broker_address="test.mosquitto.org"
broker_port=1883
```

#4. BUAT FUNCTION/ROUTINE CALLBACK UNTUK PENGIRIMAN DATA

```python
def on_publish(client,userdata,result):
    print(f"data ready bos")
    pass
```

#5. HUBUNGKAN KE BROKER MQTT

```python
client= mqtt.Client(player)                            #clientnya harus random
client.on_publish = on_publish                           #assign function to callback
client.connect(broker_address,broker_port)               #establish connection
```

#6. KIRIM DATA RANDOM KE BROKER DENGAN TOPIK YANG DIPILIH

```python
topik1 = "/gempa/getarangempadjawa"
topik2 = "/gempa/getarangempasuenda"
topik3 = "/gempa/getarangempabatak"
topik4 = "/gempa/getarangempabali"
while 1:                                                    #loop terus sambil kirim data
    getarangempadjawa = float(randrange(10,80)/10)              #random voltase dan arus
    getarangempasuenda = float(randrange(10,80)/10)
    getarangempabatak = float(randrange(10,80)/10)
    getarangempabali = float(randrange(10,80)/10)
    print(f" getaran gempa djawa = {getarangempadjawa} Magnitude , getaran gempa suenda = {getarangempasuenda} Magnitude , getaran gempa batak = {getarangempabatak} Magnitude ,getaran gempa bali= {getarangempabali} Magnitude")
    ret= client.publish(topik1,getarangempadjawa)                       #PUB suhu ke topik1 /vokasi/volt
    ret= client.publish(topik2,getarangempasuenda)
    ret= client.publish(topik3,getarangempabatak)
    ret= client.publish(topik4,getarangempabali)                #PUB suhu ke topik2 /vokasi/arus
    sleep(5)
```

#7. SILAHKAN KREASIKAN UNTUK DATA SELAIN VOLTASE DAN ARUS DIATAS
JANGAN GUNAKAN LAGI VOLT & ARUS UNTUK LAPORAN UTS, DIBAWAH INI CONTOH SAJA,
DATA LEBIH DARI 2 LEBIH BAGUS

```python
topik1 = "/vokasi/windspeed"
topik2 = "/vokasi/direction"
while 1:                                                    #loop terus sambil kirim data
    wsp = float( randrange(0,1000,2) /10  )                      #random kecepatan angin dan arah
    dir = randrange(0,360,10)
    print(f"kecepatan angin = {wsp} km/h , arah = {dir} °")
    ret= client.publish(topik1,wsp)                       #PUB suhu ke topik1 /vokasi/winspeed
    ret= client.publish(topik2,dir)                       #PUB suhu ke topik2 /vokasi/direction
    sleep(1)
```
